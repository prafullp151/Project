```python
import csv
import tensorflow as tf

from sklearn.model_selection import train_test_split

# Read data in from file
with open("banknotes.csv") as f:
    reader = csv.reader(f)
    next(reader)

    data = []
    for row in reader:
        data.append({
            "evidence": [float(cell) for cell in row[:4]],
            "label": 1 if row[4] == "0" else 0
        })

```

```python
# Separate data into training and testing groups
evidence = [row["evidence"] for row in data]
labels = [row["label"] for row in data]
X_training, X_testing, y_training, y_testing = train_test_split(
    evidence, labels, test_size=0.4
)
```

```python
# Create a neural network
model = tf.keras.models.Sequential()
```

```python
# Add a hidden layer with 8 units, with ReLU activation
model.add(tf.keras.layers.Dense(8, input_shape=(4,), activation="relu"))
```

```python
# Add a hidden layer with 8 units, with ReLU activation
model.add(tf.keras.layers.Dense(8, activation="relu"))

# Add an output layer with 1 unit for binary classification
model.add(tf.keras.layers.Dense(1, activation="sigmoid"))
```

```python
# Train neural network
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
import numpy as np

# Convert training data to numpy arrays and train
model.fit(np.array(X_training), np.array(y_training), epochs=20)
```

```python
# Evaluate how well model performs
model.evaluate(np.array(X_testing), np.array(y_testing), verbose=2)
```
